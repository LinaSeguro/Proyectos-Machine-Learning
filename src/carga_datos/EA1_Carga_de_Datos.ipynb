## 3. Transformaciones del DataFrame

### 3.1 Renombrado de columnas

Se reemplazan los nombres originales por nombres en español para facilitar la interpretación de las variables.

```python
df_espanol = df.rename(columns={
    "mpg": "millas_por_galon",
    "cylinders": "cilindros",
    "displacement": "cilindrada",
    "horsepower": "caballos_de_fuerza",
    "weight": "peso",
    "acceleration": "aceleracion",
    "model year": "anio_modelo",
    "origin": "origen",
    "car name": "nombre_automovil"
})

df_espanol.head()
```

```python
print("COLUMNAS DEL DATAFRAME")
print("-" * 30)

for numero, columna in enumerate(df_espanol.columns, start=1):
    print(f"{numero}. {columna}")
```

### 3.2 Selección de variables relevantes

Se conservan cuatro variables relevantes para analizar los automóviles: millas por galón, cilindros, caballos de fuerza y peso.

```python
df_seleccion = df_espanol[
     [
        "nombre_automovil",
        "millas_por_galon",
        "cilindros",
        "caballos_de_fuerza",
        "peso"
    ]
]

df_seleccion.head()
```

### 3.3 Filtrado por rendimiento

Se seleccionan los automóviles cuyo rendimiento es igual o superior a 25 millas por galón.

```python
df_filtrado = df_seleccion[
    df_seleccion["millas_por_galon"] >= 25
].copy()

df_filtrado[["nombre_automovil", "millas_por_galon"]].head(5)
```

```python
print(df_seleccion.columns.tolist())
```

```python
print("Filas originales:", len(df_seleccion))
print("Filas filtradas:", len(df_filtrado))
```

### 3.4 Clasificación de la eficiencia

Se crea la columna `nivel_eficiencia` y se asigna una categoría según los umbrales de rendimiento definidos en el código.

```python
df_clasificado = df_espanol.copy()

df_clasificado["nivel_eficiencia"] = df_clasificado[
    "millas_por_galon"
].apply(
    lambda x: "Alta" if x >= 30
    else "Media" if x >= 20
    else "Baja"
)

df_clasificado.sample(5)
```

## 4. Reporte final

Se genera y guarda un segundo reporte a partir del DataFrame clasificado, con el propósito de documentar su estado después de las transformaciones.

```python
ruta = reportes.guardar_reporte(
    df_clasificado,
    "reportes/reporte_auto_mpg_clasificado.txt",
)

print(f"Reporte guardado en: {ruta}")
```
